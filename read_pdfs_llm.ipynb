# Collect PDF Files 

```python
# Get all pdf files -- restricted to the Laws alone for now
pdf_path = 'Documents For RAG'+os.sep+'PDFs'+os.sep+'Laws of Cricket'

files_list = []
for root, dirs, files in os.walk(pdf_path):
    for file in files:
      files_list.append(os.path.join(root, file))

```

```python
# extract pdf as text and section data

def partition_pdf_with_footer_removal(pdf_path):#, output_dir):
    with open(pdf_path, 'rb') as file:
        reader = PyPDF2.PdfReader(file)
        num_pages = len(reader.pages)

        remove_text=rf"Laws of Cricket 2017 Code \(3rd Edition - 2022\)\d+"

        all_text = ""
        for page_num in range(num_pages):
            page = reader.pages[page_num]

            text = page.extract_text()
            text = re.sub(remove_text,"",text)
            all_text += text
            
    return all_text


section_titles = ["THE PREFACE", "CONTENTS", "THE PREAMBLE ","LAW ","APPENDIX "]
```

```python
def find_all_occurrences(text, substring):
    start = 0
    while True:
        start = text.find(substring, start)
        if start == -1:
            break
        yield start
        start += 1
```
